```python
def process_numbers():
    count = int(input("how many numbers do you want to enter? "))

    total = 0

    for i in range(count):
        num = int(input(f"enter number {i + 1}: "))

        total += num

        if num % 2 == 0:
            print(f"{num} is even")
        else:
            print(f"{num} is odd")

    if count > 0:
        average = total / count
    else:
        average = 0



    print("processing results...")




    print("\n results")
    print("total:", total)
    print("average:", average)


process_numbers()


print("THANK YOU FOR USING THE SIMPLE NUMBER EVALUATION PROGRAM")
```
